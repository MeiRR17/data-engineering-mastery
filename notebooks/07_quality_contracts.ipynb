## 1. Predict first

לפני Run All: כתוב מה אתה מצפה שה־assert יוכיח ומה אינו בודק. לאחר מכן הרץ את התא.

```python
rows=[{"id":"A","magnitude":2.1},{"id":"B","magnitude":None},
      {"id":"C","magnitude":3.4},{"id":None,"magnitude":4.4}]
valid=[x for x in rows if isinstance(x.get("id"),str) and x["id"]]
valid_ratio=len(valid)/len(rows)
completeness=sum(x["magnitude"] is not None for x in valid)/len(valid)
assert valid_ratio==0.75
assert round(completeness,3)==0.667
print("Valid:",valid_ratio,"known magnitude:",round(completeness,3))
```

## 2. Failure mode / independent check

שנה פרמטר אחד בענף האישי והסבר מדוע הקוד נכשל או משנה תוצאה. החזר את הערך המקורי.

```python
thresholds={"minimum_valid_ratio":0.75,"minimum_completeness":0.75}
publish=valid_ratio>=thresholds["minimum_valid_ratio"] and completeness>=thresholds["minimum_completeness"]
assert not publish
rejected=[{"row_index":i,"reason":"MISSING_ID"} for i,x in enumerate(rows) if not x.get("id")]
assert rejected==[{"row_index":3,"reason":"MISSING_ID"}]
print("Correct gate: block entire batch; preserve reason code only, not payload.")
```
