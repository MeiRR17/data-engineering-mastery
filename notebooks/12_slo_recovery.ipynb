## 1. Predict before running

Before selecting Run All, write down what the next assertion should prove and what it does **not** verify. Run the cell only after making your prediction.

```python
from datetime import datetime,timezone
runs=["success"]*9+["failed"]
availability=runs.count("success")/len(runs)
target=0.9
allowed_failures=int(len(runs)*(1-target)+1e-9)
assert availability==0.9
assert allowed_failures==1
print("Availability:",availability,"budget:",allowed_failures)
```
