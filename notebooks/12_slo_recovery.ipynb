## 1. Predict first

לפני Run All: כתוב מה אתה מצפה שה־assert יוכיח ומה אינו בודק. לאחר מכן הרץ את התא.

```python
from datetime import datetime,timezone
runs=["success"]*9+["failed"]
availability=runs.count("success")/len(runs)
target=0.9
allowed_failures=int(len(runs)*(1-target)+1e-9)
assert availability==0.9
assert allowed_failures==1
print("Availability:",availability,"budget:",allowed_failures)
```

## 2. Failure mode / independent check

שנה פרמטר אחד בענף האישי והסבר מדוע הקוד נכשל או משנה תוצאה. החזר את הערך המקורי.

```python
now=datetime(2026,1,3,12,tzinfo=timezone.utc)
last_event=datetime(2026,1,1,10,tzinfo=timezone.utc)
age_hours=(now-last_event).total_seconds()/3600
assert age_hours==50
freshness_limit=24
assert age_hours>freshness_limit
print("Freshness SLO breached although 9/10 runs succeeded.")
```
