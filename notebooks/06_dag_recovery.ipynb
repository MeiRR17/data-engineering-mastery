```python
from graphlib import TopologicalSorter
dependencies={"extract":set(),"validate":{"extract"},"load":{"validate"},"publish":{"load"}}
sequence=list(TopologicalSorter(dependencies).static_order())
assert sequence==["extract","validate","load","publish"]
print("DAG order:",sequence)
```

## 2. Independent failure check

On your own practice branch, change one parameter. Explain why the assertion fails or the output changes, then restore the original value. Preserve the invariant.

```python
from datetime import date,timedelta
begin,end=date(2026,1,1),date(2026,1,3)
days=[(begin+timedelta(days=i)).isoformat() for i in range((end-begin).days+1)]
assert days==["2026-01-01","2026-01-02","2026-01-03"]
failed={"2026-01-02"}
replay=[day for day in days if day in failed]
assert replay==["2026-01-02"]
print("Bounded recovery windows:",replay)
```
