## 2. Failure mode / independent check

שנה פרמטר אחד בענף האישי והסבר מדוע הקוד נכשל או משנה תוצאה. החזר את הערך המקורי.

```python
import sqlite3
db = sqlite3.connect(":memory:")
db.execute("create table training(id integer primary key, value text)")
db.execute("insert into training(value) values ('synthetic')")
db.commit()
db.execute("pragma query_only=on")
assert db.execute("select count(*) from training").fetchone()[0] == 1
try:
    db.execute("delete from training")
    raise AssertionError("read-only connection unexpectedly accepted DELETE")
except sqlite3.OperationalError:
    print("Local SQLite query-only demonstration succeeded. PostgreSQL RLS still needs a real PostgreSQL test.")
db.close()
```
