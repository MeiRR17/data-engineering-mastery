## 1. Predict first

לפני Run All: כתוב מה אתה מצפה שה־assert יוכיח ומה אינו בודק. לאחר מכן הרץ את התא.

```python
from io import BytesIO
from zipfile import ZipFile,ZipInfo,ZIP_DEFLATED
from hashlib import sha256
def artifact():
    buff=BytesIO()
    with ZipFile(buff,"w",ZIP_DEFLATED) as z:
        info=ZipInfo("app.txt",(1980,1,1,0,0,0))
        info.compress_type=ZIP_DEFLATED
        z.writestr(info,"synthetic-training-version-1")
    return buff.getvalue()
a,b=artifact(),artifact()
assert a==b
digest=sha256(a).hexdigest()
print("Deterministic demo ZIP SHA-256:",digest[:16])
```

## 2. Failure mode / independent check

שנה פרמטר אחד בענף האישי והסבר מדוע הקוד נכשל או משנה תוצאה. החזר את הערך המקורי.

```python
tamp=a+b"unauthorized bytes"
assert sha256(tamp).hexdigest()!=digest
active="v1"
candidate="v2"
health="unhealthy"
if health=="ready":
    active=candidate
assert active=="v1"
print("Never switch the active version after failed health or integrity checks.")
```
