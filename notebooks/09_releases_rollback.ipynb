## 1. Predict before running

Before selecting Run All, write down what the next assertion should prove and what it does **not** verify. Run the cell only after making your prediction.

```python
from io import BytesIO
from zipfile import ZipFile,ZipInfo,ZIP_DEFLATED
from hashlib import sha256
def artifact():
    buff=BytesIO()
    with ZipFile(buff,"w",ZIP_DEFLATED) as z:
        info=ZipInfo("app.txt",(1980,1,1,0,0,0))
        info.compress_type=ZIP_DEFLATED
        z.writestr(info,"synthetic-training-version-1")
    return buff.getvalue()
a,b=artifact(),artifact()
assert a==b
digest=sha256(a).hexdigest()
print("Deterministic demo ZIP SHA-256:",digest[:16])
```
